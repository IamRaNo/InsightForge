## Basic ***Data Preprocessing***

````python
data.columns = data.columns.str.lower()
````

````python
data.drop(columns="report number",inplace=True)
````

````python
data.head()
````

````python
rename_dict = {'date reported':'report_date','date of occurrence':'crime_date','time of occurrence':'crime_time','crime code':'crime_code',
               'crime description':'description',
               'victim age':'victim_age','victim gender':'victim_gender','weapon used':'weapon','crime domain':'crime_domain','police deployed':'police_deployed',
               'case closed':'case_close_status','date case closed':'close_date'}


data.rename(columns=rename_dict,inplace=True)
````

````python
data.sample(5)
````

````python
data['report_date']=pd.to_datetime(data['report_date'],format='%d-%m-%Y %H:%M')
````

````python
data['crime_date']=pd.to_datetime(data['crime_date'],format='%m-%d-%Y %H:%M')
````

````python
data['crime_time']=pd.to_datetime(data['crime_time'],format='%d-%m-%Y %H:%M')
````

````python
data['close_date']=pd.to_datetime(data['close_date'],format='%d-%m-%Y %H:%M')
````

## ***Categorical*** Columns

**City**

````python
data['city']=data['city'].apply(lambda x: x.lower())
````

**Gender**

````python
data['victim_gender'].value_counts()
````

````python
gender_dict ={
    'F':'Female',
    'M':'Male',
    'X':'Unknown'
}


data['victim_gender']=data['victim_gender'].map(gender_dict)
````

````python
data['victim_gender'].value_counts()
````

**Description**

````python
data['description'].value_counts()
````

````python
data['description']=data['description'].apply(lambda x: x.title())
````
